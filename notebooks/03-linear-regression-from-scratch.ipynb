# Linear Regression From Scratch

```python
import numpy as np

rng = np.random.default_rng(0)
x = rng.normal(size=100)
y = 2.5 * x + 0.7 + rng.normal(scale=0.2, size=100)
w, b = 0.0, 0.0
lr = 0.1
for step in range(200):
    pred = w * x + b
    error = pred - y
    w -= lr * (2 * (error * x).mean())
    b -= lr * (2 * error.mean())
print(round(w, 3), round(b, 3))
```
