```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = {
    'vacation': 'Employees get 20 vacation days per year.',
    'expenses': 'Expenses must be submitted within 30 days.',
    'security': 'Passwords must use multi-factor authentication.',
}
question = 'How many days off do employees get?'
vectorizer = TfidfVectorizer().fit(list(docs.values()) + [question])
scores = cosine_similarity(vectorizer.transform([question]), vectorizer.transform(docs.values()))[0]
best_key = list(docs)[scores.argmax()]
print('Retrieved:', best_key, docs[best_key])
print('Answer:', docs[best_key])
```
