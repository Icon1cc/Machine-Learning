# Pandas EDA

Practice loading, summarizing, and grouping tabular data.

```python
import pandas as pd

df = pd.DataFrame({
    'plan': ['free', 'pro', 'free', 'team', 'pro'],
    'usage': [3, 18, 5, 31, 21],
    'churned': [1, 0, 1, 0, 0],
})
print(df)
print(df.groupby('plan').agg({'usage': 'mean', 'churned': 'mean'}))
```
