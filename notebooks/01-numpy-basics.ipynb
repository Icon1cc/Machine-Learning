# NumPy Basics

Learn arrays, vectorization, and reproducible random numbers.

```python
import numpy as np

rng = np.random.default_rng(42)
x = rng.normal(size=(5, 3))
print(x)
print('column means:', x.mean(axis=0))
```

```python
weights = np.array([0.2, -0.5, 1.0])
scores = x @ weights
print(scores)
```
