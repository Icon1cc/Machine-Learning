# Embeddings and Semantic Search

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = ['refund policy for annual plan', 'reset password instructions', 'invoice and billing help']
query = 'how do I get my money back'
vectorizer = TfidfVectorizer().fit(docs + [query])
S = cosine_similarity(vectorizer.transform([query]), vectorizer.transform(docs))[0]
print(sorted(zip(S, docs), reverse=True))
```
