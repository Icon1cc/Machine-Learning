# Neural Network From Scratch

```python
import numpy as np

rng = np.random.default_rng(2)
X = rng.normal(size=(300, 2))
y = (X[:, 0] * X[:, 1] > 0).astype(float).reshape(-1, 1)
W1 = rng.normal(scale=0.5, size=(2, 4)); b1 = np.zeros((1, 4))
W2 = rng.normal(scale=0.5, size=(4, 1)); b2 = np.zeros((1, 1))
for step in range(1000):
    h = np.tanh(X @ W1 + b1)
    p = 1 / (1 + np.exp(-(h @ W2 + b2)))
    dz2 = p - y
    W2 -= 0.1 * h.T @ dz2 / len(X); b2 -= 0.1 * dz2.mean(axis=0)
    dz1 = (dz2 @ W2.T) * (1 - h ** 2)
    W1 -= 0.1 * X.T @ dz1 / len(X); b1 -= 0.1 * dz1.mean(axis=0)
print('accuracy:', ((p > 0.5) == y).mean())
```
